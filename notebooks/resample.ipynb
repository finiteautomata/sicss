```python
import json
from datasets import load_dataset

# Load non-raw dataset and build lookup by id
ds = load_dataset('piuba-bigdata/contextualized_hate_speech', split='train')
lookup = {r['id']: r for r in ds}
print(f"Non-raw dataset size: {len(lookup)}")

# Load sample.json, enrich, and write back
enriched = []
missing = 0
with open('sample.json') as f:
    for line in f:
        rec = json.loads(line)
        nid = rec['id']
        if nid in lookup:
            non_raw = lookup[nid]
            rec['context_tweet'] = non_raw.get('context_tweet', '')
            rec['title'] = non_raw.get('title', '')
            rec['body'] = non_raw.get('body', '')
        else:
            print(f"  No match for id={nid}, tweet_id={rec.get('tweet_id')}")
            missing += 1
        enriched.append(rec)

print(f"Total records: {len(enriched)}, missing matches: {missing}")

# Write back
with open('sample.json', 'w') as f:
    for rec in enriched:
        f.write(json.dumps(rec, ensure_ascii=False) + '\n')

print("Done.")
EOF
```
